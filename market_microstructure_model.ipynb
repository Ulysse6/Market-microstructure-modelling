# Market Microstructure Model
Simple simulation of order flow, bid-ask spread and market impact.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
N=5000
midprice0=100.0
spread=0.02
impact=0.005
np.random.seed(42)
```

```python
order_sign=np.where(np.random.rand(N)<0.55,1,-1)
order_size=np.random.exponential(1.0,N)
```

```python
mid=np.zeros(N+1)
mid[0]=midprice0
for t in range(N):
    mid[t+1]=mid[t]+impact*order_sign[t]*order_size[t]+0.001*np.random.randn()

bid=mid[:-1]-spread/2
ask=mid[:-1]+spread/2
trade=np.where(order_sign>0,ask,bid)
```

```python
plt.figure(figsize=(10,4))
plt.plot(mid)
plt.title('Midprice')
plt.show()
```

```python
imbalance=pd.Series(order_sign).rolling(100).mean()
imbalance.plot(figsize=(10,4),title='Order Flow Imbalance')
```
